## Parte I — k-means

### 1. Implementação

```python
def kminit(X, K, flag):
    # Inicialização - aleatória (flag = 0) ou k-means++ (flag = 1)
    # A inicialização do primeiro centróide é fixa
    N = X.shape[0]
    C = []
    init_centr_idx = np.random.randint(N)
    C.append(X[init_centr_idx])

    if flag == 0:   # random
        idx = np.random.choice(N, K - 1, replace=False)
        while init_centr_idx in idx:   # caso raro --> repetiu o primeiro
            idx = np.random.choice(N, K - 1, replace=False)
        C.extend(X[idx])

    else:           # k-means++
        for _ in range(K - 1):
            C_atual = np.array(C)
            M = np.zeros((N, len(C_atual)))

            # distância ao quadrado de cada pose a cada centroide já escolhido
            for k in range(len(C_atual)):
                M[:, k] = np.sum((X - C_atual[k])**2, axis=1)

            # distância de cada pose ao centroide MAIS PRÓXIMO: (N,)
            d2 = np.min(M, axis=1)

            # probabilidade proporcional a d²
            prob = d2 / np.sum(d2)

            new_idx = np.random.choice(N, p=prob)
            C.append(X[new_idx])

    return np.array(C)
```

```python
def kmeans_custo(X, C, s):
    # Calcula a função de custo do k-means
    centr_point = C[s]                                   # centroide de cada pose: (N, 66)
    dist_eucl = np.sum((X - centr_point)**2, axis=1)     # distância ao quadrado de cada pose ao seu centroide: (N,)
    total_cost = np.sum(dist_eucl)
    return total_cost
```

```python
def kmeans(X, K, flag):
    # k-means sobre a matriz de dados X
    C = kminit(X, K, flag)
    prev_cost = np.inf
    N = X.shape[0]

    while True:
        M = np.zeros((N, K))

        for k in range(K):
            M[:, k] = np.sum((X - C[k])**2, axis=1)     # distância ao quadrado de cada pose ao centroide k: (N,)

        s = np.argmin(M, axis=1)                        # para cada pose, o centroide mais próximo

        total_cost = kmeans_custo(X, C, s)

        # Comparacao dos custos
        if prev_cost - total_cost < 1e-6:
            break
        prev_cost = total_cost

        # Atualizacao dos centroides
        for k in range(K):
            cluster = X[s == k]
            if len(cluster) > 0:
                C[k] = np.mean(cluster, axis=0)

    return C, s
```

### 6. Redução de dimensionalidade — PCA implementado com numpy (sem scikit-learn)

```python
# Pergunta 6 - funções implementadas

def pca_projetar(X, r):
    # Projeta X (N, d) nas r primeiras componentes principais.
    # Devolve Z (N, r), W (d, r), a média mu (d,) e todos os valores próprios por ordem decrescente.
    N = X.shape[0]
    mu = np.mean(X, axis=0)
    Xc = X - mu
    Sigma = (Xc.T @ Xc) / N                 # covariância (d, d)

    lambdas, V = np.linalg.eigh(Sigma)      # eigh porque Sigma é simétrica
    ordem = np.argsort(lambdas)[::-1]       # eigh devolve por ordem crescente
    lambdas = lambdas[ordem]
    V = V[:, ordem]

    W = V[:, :r]
    Z = Xc @ W
    return Z, W, mu, lambdas


def pca_reconstruir(Z, W, mu):
    # Volta à dimensão original: X ≈ Z W^T + mu
    return Z @ W.T + mu


def variancia_explicada(lambdas, r):
    return np.sum(lambdas[:r]) / np.sum(lambdas)
```

```python
# Verificação: comparar a implementação própria com o PCA do scikit-learn (Pergunta 5)
Z_chk, W_chk, mu_chk, lam_chk = pca_projetar(X_norm, 66)
pca_sk = PCA().fit(X_norm)

print(np.allclose(lam_chk / lam_chk.sum(), pca_sk.explained_variance_ratio_))
print(np.allclose(np.abs(Z_chk), np.abs(pca_sk.transform(X_norm)), atol=1e-6))   # igual a menos do sinal
print(np.mean((X_norm - pca_reconstruir(Z_chk, W_chk, mu_chk))**2))              # r = 66 -> erro ~0
```

```python
# Funções auxiliares para R6.b e R6.c
def avaliar_reconstrucao(r):
    Z, W, mu, lambdas = pca_projetar(X_norm, r)
    X_rec_norm = pca_reconstruir(Z, W, mu)

    # erro em coordenadas originais (u, v)
    X_rec = scaler.inverse_transform(X_rec_norm)

    var_exp  = variancia_explicada(lambdas, r)
    mse_norm = np.mean((X_norm - X_rec_norm)**2)       # espaço normalizado
    mse_orig = np.mean((X - X_rec)**2)                 # coordenadas originais

    # erro por marcador: (N, 66) -> (N, 33, 2) -> média sobre poses e sobre (u, v)
    erro_marc = np.mean(((X - X_rec)**2).reshape(-1, 33, 2), axis=(0, 2))

    return Z, X_rec, var_exp, mse_norm, mse_orig, erro_marc


def mostrar_resultados(r, var_exp, mse_norm, mse_orig, erro_marc, top=8):
    print(f"r = {r}")
    print(f"  Variância explicada:            {var_exp*100:.2f} %")
    print(f"  MSE (espaço normalizado):       {mse_norm:.4f}")
    print(f"  MSE (coordenadas originais):    {mse_orig:.6f}")
    print(f"  Marcadores com maior erro:")
    for j in np.argsort(erro_marc)[::-1][:top]:
        print(f"    {j:2d} {NAME_MARKERS[j]:<13s} {erro_marc[j]:.6f}")
```

### R7.b)

```python
from matplotlib.patches import Ellipse

def desenhar_elipse(ax, media, cov, n_std=2, **kwargs):
    valores, vetores = np.linalg.eigh(cov)          # valores por ordem crescente
    angulo = np.degrees(np.arctan2(vetores[1, -1], vetores[0, -1]))
    largura = 2 * n_std * np.sqrt(valores[-1])      # eixo maior
    altura  = 2 * n_std * np.sqrt(valores[0])       # eixo menor
    ax.add_patch(Ellipse(media, largura, altura, angle=angulo, **kwargs))

K_gmm = 4
gmm = GaussianMixture(n_components=K_gmm, covariance_type='full',
                      n_init=10, random_state=0).fit(Z95)
labels_gmm = gmm.predict(Z95)
print(gmm.weights_.round(3))

colors = plt.cm.tab10(np.arange(K_gmm))
fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(Z95[:, 0], Z95[:, 1], c=colors[labels_gmm], s=4, alpha=0.3)

for k in range(K_gmm):
    media = gmm.means_[k, :2]
    cov = gmm.covariances_[k, :2, :2]
    desenhar_elipse(ax, media, cov, n_std=2, edgecolor=colors[k], facecolor='none', lw=2)
    ax.scatter(*media, color=colors[k], marker='X', s=200, edgecolor='black', label=f'Componente {k}')

ax.set(xlabel='PC1', ylabel='PC2', title='GMM (K = 4) na projeção 2D')
ax.legend(); ax.grid(alpha=0.3)
plt.show()

for k in range(K_gmm):
    print(f'Componente {k}:')
    print(gmm.covariances_[k, :2, :2].round(2))
```

```python
K_poses = 11
gmm_poses = GaussianMixture(n_components=K_poses, covariance_type='full', n_init=5, random_state=0).fit(Z95)
vect_clust = gmm_poses.predict(Z95)

H = []
start = 0

for seq in df_train['Skeleton_Sequence']:
    T = len(seq)                                   # nº de poses desta sequência
    label = vect_clust[start:start+T]              # clusters das poses desta sequência
    H.append(np.bincount(label, minlength=K_poses) / T)
    start += T

H = np.array(H)                                     # (71, K_poses)
```

```python
Ks2 = list(range(2, 16))

custos_H = np.zeros(len(Ks2))
sil_H    = np.zeros(len(Ks2))
melhores = {}

for i, K in enumerate(Ks2):
    best_custo, best = np.inf, None
    for _ in range(20):                          # 20 execuções, guarda a de menor custo
        C, s = kmeans(H, K, 1)
        c = kmeans_custo(H, C, s)
        if c < best_custo:
            best_custo, best = c, (C, s)
    melhores[K] = best
    custos_H[i] = best_custo
    sil_H[i]    = silhouette_score(H, best[1])
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].plot(Ks2, custos_H, 'o-')
ax[0].set(xlabel='Número de exercícios K₂', ylabel='Custo k-means', title='Método do cotovelo (sequências)')
ax[1].plot(Ks2, sil_H, 'o-')
ax[1].set(xlabel='Número de exercícios K₂', ylabel='Silhouette score', title='Silhouette (sequências)')
for a in ax:
    a.set_xticks(list(Ks2)); a.grid(alpha=0.3)
plt.savefig('R9)_K=11.jpg', dpi=150, bbox_inches='tight')
plt.show()
```
