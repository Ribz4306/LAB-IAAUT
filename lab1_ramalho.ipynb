## Imports e seed

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# NAO MODIFICAR A SEED
# flag = 0 --> random
# flag = 1 --> km++
np.random.seed(555)
```

## Leitura e normalização dos dados

```python
df_train = pd.read_pickle('Xtrain.pkl')

# todas as poses de todas as sequências numa única matriz: (8374, 66)
X = np.concatenate(df_train['Skeleton_Sequence'].to_numpy())
print(X.shape)

# Normalização dos dados - centrar + escalamento (recomendado e crítico para o PCA)
scaler = StandardScaler()
X_norm = scaler.fit_transform(X)
```

## Parte I — k-means

### 1. Implementação

```python
def kminit(X, K, flag):
    # Inicialização - aleatória (flag = 0) ou k-means++ (flag = 1)
    # A inicialização do primeiro centróide é fixa
    N = X.shape[0]
    C = []
    init_centr_idx = np.random.randint(N)
    C.append(X[init_centr_idx])

    if flag == 0:   # random
        idx = np.random.choice(N, K - 1, replace=False)
        while init_centr_idx in idx:   # caso raro --> repetiu o primeiro
            idx = np.random.choice(N, K - 1, replace=False)
        C.extend(X[idx])

    else:           # k-means++
        for _ in range(K - 1):
            C_atual = np.array(C)
            M = np.zeros((N, len(C_atual)))

            # distância ao quadrado de cada pose a cada centroide já escolhido
            for k in range(len(C_atual)):
                M[:, k] = np.sum((X - C_atual[k])**2, axis=1)

            # distância de cada pose ao centroide MAIS PRÓXIMO: (N,)
            d2 = np.min(M, axis=1)

            # probabilidade proporcional a d²
            prob = d2 / np.sum(d2)

            new_idx = np.random.choice(N, p=prob)
            C.append(X[new_idx])

    return np.array(C)
```

```python
def kmeans_custo(X, C, s):
    # Calcula a função de custo do k-means
    centr_point = C[s]                                   # centroide de cada pose: (N, 66)
    dist_eucl = np.sum((X - centr_point)**2, axis=1)     # distância ao quadrado de cada pose ao seu centroide: (N,)
    total_cost = np.sum(dist_eucl)
    return total_cost
```

```python
def kmeans(X, K, flag):
    # Aplica o algoritm k-means a matriz de dados X.
    C = kminit(X,K,flag)
    prev_cost = np.inf
    N = X.shape[0]

    while True:
        M = np.zeros((N, K))

        for k in range(K):
            M[:, k] = np.sum((X - C[k])**2, axis=1)     # distância ao quadrado de cada pose ao centroide k: (N,)

        s = np.argmin(M, axis=1)                        # para cada pose, o centroide mais próximo

        total_cost = kmeans_custo(X, C, s)

        # Comparacao dos custos
        if prev_cost - total_cost < 1e-6:
            break
        prev_cost = total_cost

        # Atualizacao dos centroides
        for k in range(K):
            cluster = X[s == k]
            if len(cluster) > 0:
                C[k] = np.mean(cluster, axis=0)

    return C, s
```

### 2. Selecionar o número de centróides e inicialização

#### R2.a) Inicialização aleatória — mínimo de 10 execuções por K

```python
Ks = list(range(1, 21))
n_runs = 10

custos_a = np.zeros((len(Ks), n_runs))   # linha = K, coluna = execução
for i, K in enumerate(Ks):
    for r in range(n_runs):
        C, s = kmeans(X_norm, K, 0)
        custos_a[i, r] = kmeans_custo(X_norm, C, s)
    print(f"K = {K} feito")

min_random = np.min(custos_a, axis=1)
```

```python
plt.plot(Ks, min_random, 'o-', label='random')
plt.axvline(x=3, color='red', linestyle='--', label='K = 3')
plt.xlabel('Número de centroides K')
plt.ylabel('Custo')
plt.xticks(Ks)
plt.legend()
plt.grid(True)
plt.savefig('R2a.png', dpi=150)
plt.show()

descidas = -np.diff(min_random)
print("Descidas entre K consecutivos:", descidas)
print("Variação do declive:", -np.diff(descidas))
```

#### R3.a) Centroides do melhor modelo (K = 3, k-means++, melhor de 10 execuções)

```python
CONNECTIONS = [
    (0,1),(1,2),(2,3),(3,7),(0,4),(4,5),(5,6),(6,8),(9,10),      # cara
    (11,12),(11,23),(12,24),(23,24),                              # tronco
    (11,13),(13,15),(15,17),(15,19),(15,21),(17,19),              # braço esquerdo
    (12,14),(14,16),(16,18),(16,20),(16,22),(18,20),              # braço direito
    (23,25),(25,27),(27,29),(27,31),(29,31),                      # perna esquerda
    (24,26),(26,28),(28,30),(28,32),(30,32),                      # perna direita
]

def desenhar_pose(ax, pose, title):
    p = pose.reshape(33, 2)                  # 33 marcadores × (u, v)
    for a, b in CONNECTIONS:
        ax.plot([p[a, 0], p[b, 0]], [p[a, 1], p[b, 1]], 'b-')
    ax.scatter(p[:, 0], p[:, 1], c='red', s=15, zorder=3)
    ax.invert_yaxis()
    ax.set_aspect('equal')
    ax.set_title(title)
```

### 6. Redução de dimensionalidade — PCA implementado com numpy


O PCA procura as direções em que as poses mais variam em relação à pose média:
1. **Centrar:** subtrair a pose média μ a todas as poses.
2. **Covariância (66×66):** diz que coordenadas variam em conjunto.
3. **Vetores e valores próprios da covariância:** os vetores são as componentes principais (direções de movimento) e os valores próprios são a variância ao longo de cada uma.
4. **Projetar:** Z = Xc · W, ficando cada pose descrita por apenas r números (W = as r primeiras componentes).
5. **Reconstruir:** X̂ = Z · Wᵀ + μ.

```python
def pca_project(X, r):
    """
    Projeta X (N, d) em r componentes principais.
    Devolve: Z (N, r) projeção, W (d, r) componentes (uma por coluna),
             mu (d,) pose média, lambdas (d,) variância de todas as componentes (decrescente).
    """
    mu = X.mean(axis=0)                         
    Xc = X - mu                                 
    cov_matrix = Xc.T @ Xc / X.shape[0]         
    lambdas, V = np.linalg.eigh(cov_matrix)     
    lambdas, V = lambdas[::-1], V[:, ::-1]      
    W = V[:, :r]                                
    Z = Xc @ W                                  
    return Z, W, mu, lambdas


def pca_reconstruct(Z, W, mu):
    """Reconstrói X (N, d) a partir da baixa dimensão: X_rec = Z @ W.T + mu."""
    return Z @ W.T + mu


def reconstruction_error(X, X_rec):
    """MSE global e MSE por marcador (33 valores, média sobre poses e sobre u, v)."""
    e = (X - X_rec) ** 2
    return e.mean(), e.reshape(-1, 33, 2).mean(axis=(0, 2))
```

#### R6.a) r = 2: projeção e visualização

```python
Z2, W2, mu, lambdas = pca_project(X_norm, 2)    # lambdas = variância de todas as componentes

plt.figure(figsize=(7, 5.5))
plt.scatter(Z2[:, 0], Z2[:, 1], s=3, alpha=0.3)
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.title('Poses projetadas em 2D (r = 2)')
plt.grid(True, alpha=0.3)
plt.savefig('R6a.png', dpi=150)
plt.show()
```

#### R6.b) Variância explicada e erro de reconstrução (r = 2)

```python
# reconstruir e voltar às coordenadas originais (u, v), para o erro ter significado físico
X_rec2 = scaler.inverse_transform(pca_reconstruct(Z2, W2, mu))

var_exp_2 = lambdas[:2].sum() / lambdas.sum()    # variância explicada pelas 2 componentes
mse_2, marker_err_2 = reconstruction_error(X, X_rec2)

plt.figure(figsize=(10, 4))
plt.bar(range(33), marker_err_2)
plt.xticks(range(33))
plt.xlabel('Marcador')
plt.ylabel('Erro quadrático médio')
plt.title('Erro de reconstrução por marcador (r = 2)')
plt.grid(True, axis='y', alpha=0.3)
plt.savefig('R6b.png', dpi=150)
plt.show()
```

#### R6.c) r para 95% da variância e comparação com r = 2

```python
r95 = np.argmax(np.cumsum(lambdas) / lambdas.sum() >= 0.95) + 1     # menor r com >= 95% da variância

Z95, W95, mu, lambdas = pca_project(X_norm, r95)
X_rec95 = scaler.inverse_transform(pca_reconstruct(Z95, W95, mu))

var_exp_95 = lambdas[:r95].sum() / lambdas.sum()
mse_95, marker_err_95 = reconstruction_error(X, X_rec95)

x = np.arange(33)
plt.figure(figsize=(10, 4))
plt.bar(x - 0.2, marker_err_2, width=0.4, label='r = 2')
plt.bar(x + 0.2, marker_err_95, width=0.4, label=f'r = {r95}')
plt.xticks(x)
plt.xlabel('Marcador')
plt.ylabel('Erro quadrático médio')
plt.title('Erro de reconstrução por marcador')
plt.legend()
plt.grid(True, axis='y', alpha=0.3)
plt.savefig('R6c.png', dpi=150)
plt.show()
```
